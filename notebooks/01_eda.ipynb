# Medical Cost Prediction — Exploratory Data Analysis (EDA)

## Objective
the goal of this notebook is to explore the medical insurance dataset to understand ket factors affecting medical charges and prepare the data for machine learning modeling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/ML-IA/medical-cost-prediction/data/insurance.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.hist(figsize=(10,8), bins=30)
plt.suptitle("Features Distributions")
plt.show()
```

```python
sns.histplot(df["charges"], kde=True)
```

```python
plt.Figure(figsize=(8,5))
sns.boxplot(x="smoker", y="charges", data=df)
plt.title("Smoker vs Medical charges")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="age",y="charges",data=df)
plt.title("Age vs Medical Charges")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="bmi",y="charges",data=df)
plt.title("bmi vs Medical Charges")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("correlation Matrix")
plt.show()
```

## Final EDA Insights

- Medical charges are highly right-skewed, with most people having low to moderate costs and a few extreme high-cost cases.

- Smoking is the strongest factor affecting medical charges, with smokers paying significantly higher costs.

- Age shows a positive relationship with medical charges, but the relationship is not strong enough alone for prediction.

- BMI has a moderate influence, where higher BMI tends to be associated with higher charges.

- Region and gender have minimal impact on medical charges.

- The dataset shows that multiple features together are needed for accurate prediction.
